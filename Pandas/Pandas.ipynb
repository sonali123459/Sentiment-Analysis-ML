```python
# min,max,sum total bill w.r.t each sex
pd.pivot_table(data=tip,index=['sex'],values=['total_bill'],aggfunc=['mean','max','min'])
```

```python
pd.crosstab(
    tip['day'],
    tip['sex'],
    values=tip['total_bill'],
    aggfunc='mean')
```

```python
tip.groupby("day")["tip"].sum()
```

```python
group=tip.groupby(["day","smoker"])["tip"].sum()
group
```
